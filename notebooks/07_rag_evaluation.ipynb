# 07 RAG Evaluation
Création de la table rag_evaluation et lecture du dataset de référence.

```python
create_sql = '''
CREATE TABLE IF NOT EXISTS fr_raise.rag_pipeline.rag_evaluation (
    evaluation_id STRING,
    executed_at TIMESTAMP,
    dataset_name STRING,
    question_id STRING,
    question STRING,
    expected_category STRING,
    predicted_category STRING,
    reference_answer STRING,
    answer STRING,
    sources_json STRING,
    categories_json STRING,
    retrieval_count INT,
    retrieval_success INT,
    response_time_ms DOUBLE,
    relevance_score DOUBLE,
    faithfulness_score DOUBLE,
    hallucination_rate DOUBLE,
    answer_quality DOUBLE,
    compliance_score DOUBLE,
    created_at TIMESTAMP
)
USING DELTA
'''
spark.sql(create_sql)
```

```python
display(spark.sql('SELECT * FROM fr_raise.rag_pipeline.rag_evaluation ORDER BY executed_at DESC LIMIT 50'))
```
